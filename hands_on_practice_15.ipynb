<b>Yield & Irrigation Efficiency Analysis for Maize Farming</b> <br>
<b>Scenario: Agriculture</b> <br>
You are working as an agricultural analyst for a commercial farm. The management team wants to evaluate daily soil moisture loss across a 10-day period and calculate the required top-up to prevent heat stress in a maize field. <br>

<b>Objective</b> <br>
This project was undertaken to help farm manager get an actionable irrigation schedule to improve maize plant health and avoid over application of irrigation water which leads to leaching of soil nutrients.

<b>Numerical Dataset</b> <br>
days: A sequence of ten numerical data <br>
temperature: An array of ten temperature records in Celsius <br>
moisture: An array of ten moisture levels in percentage <br>

<b>Calculations</b> <br>
Calculate the following: <br>
Daily moisture drop (Evaporation rate) <br>
Moisture deficit relative to the ideal 35% threshold <br>
Irrigation volume <br>
Hottest day



```python
# Code:

# Ten days record:
days = np.arange(1, 11)

# Daily temperature records in Celsius:
temperatures = np.array([28.5, 30.0, 31.5, 33.0, 34.5, 35.0, 34.0, 32.5, 31.0, 29.0])

# Soil moisture levels in % before daily evaporation
moisture_array = np.array([42.0, 40.5, 38.0, 35.0, 31.0, 27.5, 25.0, 24.0, 25.5, 28.0])


# Calculations:
# 1. Daily moisture drop (Daily evaporation rate)
moisture_drop = np.subtract(moisture_array[:-1], moisture_array[1:])


# 2. Moisture deficit relative to the ideal 35%
# The negative values are greater than the ideal moisture level and as such, maize plants don't require any 
# extra irrigation as the soil is already saturated
# Because the negative values require no irrigation, i will make them display as zero

ideal_moisture = 35
deficit = np.subtract(ideal_moisture, moisture_array)
moisture_deficit = np.maximum(0, deficit)


# 3. Irrigation volume
# This is the volume of irrigation water needed based on how far below the ideal moisture level the soil is
# Assuming a 1% drop in moisture requires a 1L of water for irrigation replenishment,
irrigation_volume = np.multiply(moisture_deficit, 1)


# 4. Hottest Day
# The single hottest day is the max data entry for temperture
hottest_day = np.max(temperatures)
# hottest_day_index gives the index of the day with the highest temperature
hottest_day_index = np.argmax(temperatures)


# Result:
print("================================================================================================")
print("                                DATA SUMMARY")
print("================================================================================================")

print("Daily Moisture Drop (%):", moisture_drop)
print()
print("Moisture Deficit below 35% (%):", moisture_deficit)
print()
print("Irrigation Volume Needed (Liters):", irrigation_volume)
print()
print(f"The hottest temperature record was {hottest_day:.0f}°C on day {hottest_day_index + 1}")

print("================================================================================================")

```

<b>Interpretation</b> <br>
Using the ten-day data for temperature and soil moisture in a maize farm, this mini-project highlights the drop in soil moisture across the days of moisture data collection, the days where soil moisture fell below the threshold (35%) and showed the hottest day in the ten day window. It also used the data gotten as moisture deficit to infer the approximate volume of irrigation water required by the maize plants using some assumptions (That 1% below moisture threshold correlates with 1L of irrigation water requirement). <br>

<b>Conclusion</b> <br>
With the output of this project, management can get valuable insights as the to the volume of irrigation water required by plants over a heatwave period. This can serve as an actionable irrigation schedule to ensure optimum plant health and prevent heat stress that can inhibit normal physiological and photosynthetic processes.
